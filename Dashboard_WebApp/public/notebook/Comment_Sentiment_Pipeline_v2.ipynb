```python
# make sure clean_text is string
df["clean_text"] = df["clean_text"].fillna("").astype(str).str.strip()

def remove_mentions(text):
    return re.sub(r"@\S+", "", text).strip()


def has_text(text):
    """
    Check whether comment contains at least
    one letter or number.
    """
    return any(char.isalnum() for char in text)


def classify_basic(text):

    text_no_mention = remove_mentions(text)
    text_lower = text_no_mention.lower().strip()

    # -----------------------------------------
    # 1. emoji / punctuation / mention only
    # -----------------------------------------

    if not has_text(text_no_mention):
        return "irrelevant_noise"


    # -----------------------------------------
    # 2. obvious conversational noise
    # -----------------------------------------

    noise_words = {
        "thanks",
        "thank you",
        "thank u",
        "tq",
        "tqtq",
        "first",
        "im first",
        "i'm first",
        "hello",
        "hi",
        "please",
        "welcome",
        "ok",
        "okay",
        "agreed",
        "sure",
        "haha",
        "hahaha"
    }

    # Remove punctuation/emojis for comparison
    simple_text = re.sub(
        r"[^a-zA-Z0-9\s']",
        "",
        text_lower
    ).strip()

    if simple_text in noise_words:
        return "irrelevant_noise"


    # -----------------------------------------
    # 3. tourism related keywords
    # -----------------------------------------

    tourism_keywords = [

        # Malaysia / destinations
        "malaysia",
        "malaysian",
        "kuala lumpur",
        "penang",
        "langkawi",
        "melaka",
        "malacca",
        "johor",
        "sabah",
        "sarawak",
        "borneo",
        "putrajaya",
        "sipadan",
        "tioman",
        "perhentian",
        "genting",
        "cameron highlands",
        "kinabalu",
        "semporna",
        "kuching",

        # Tourism
        "travel",
        "travelling",
        "traveling",
        "tour",
        "tourist",
        "tourism",
        "trip",
        "visit",
        "visited",
        "vlog",

        # Nature / attractions
        "beach",
        "island",
        "mountain",
        "park",
        "view",
        "views",
        "scenery",
        "waterfall",
        "nature",
        "diving",
        "snorkeling",
        "sea",
        "coast",
        "wildlife",

        # Accommodation / transport
        "hotel",
        "hostel",
        "resort",
        "airport",
        "flight",
        "train",
        "bus",
        "grab",
        "taxi",

        # Cost
        "price",
        "cheap",
        "expensive",
        "ringgit",

        # Food
        "food",
        "restaurant",
        "nasi lemak",
        "delicious",
        "spicy",

        # Malay / Indonesian terms
        "tempat",
        "cantik",
        "cantek",
        "menarik",
        "sedap",
        "makanan",
        "mahal",
        "murah",
        "pantai",
        "pulau",
        "pelancong",
        "wisata",
        "indah"
    ]

    for keyword in tourism_keywords:

        if keyword in text_lower:
            return "tourism_relevant"


    # -----------------------------------------
    # 4. generic sentiment
    # -----------------------------------------

    sentiment_words = [
        "good",
        "great",
        "beautiful",
        "amazing",
        "awesome",
        "bad",
        "terrible",
        "love",
        "hate",
        "best",
        "worst",
        "perfect",
        "underrated",
        "overrated"
    ]

    for word in sentiment_words:

        if re.search(r"\b" + re.escape(word) + r"\b", text_lower):
            return "sentiment_only"


    # -----------------------------------------
    # 5. can't decide
    # -----------------------------------------

    return "uncertain"


df["relevance"] = df["clean_text"].apply(classify_basic)

df
```

```python
import pandas as pd
import numpy as np
import re

# ============================================================
# 1. LOAD DATA
# ============================================================

INPUT_FILE = "v1.csv"

df = pd.read_csv(INPUT_FILE)

print("Original dataset:", df.shape)
print(df.columns.tolist())


# ============================================================
# 2. BASIC CLEANING
# ============================================================

df["clean_text"] = (
    df["clean_text"]
    .fillna("")
    .astype(str)
    .str.replace(r"\s+", " ", regex=True)
    .str.strip()
)

df["comment_text"] = (
    df["comment_text"]
    .fillna("")
    .astype(str)
)


# ============================================================
# 3. BETTER WORD COUNT
# ============================================================
# Counts actual text-like words rather than emoji chunks.

df["word_count_new"] = (
    df["clean_text"]
    .str.findall(r"\b[\w']+\b")
    .str.len()
)


# ============================================================
# 4. HELPER FUNCTIONS
# ============================================================

def normalize(text):
    """
    Lowercase + normalize whitespace.
    """
    text = str(text).lower()
    text = re.sub(r"\s+", " ", text)
    return text.strip()


def contains_any(text, keywords):
    """
    Returns True if any keyword exists in text.
    """
    text = normalize(text)

    return any(keyword.lower() in text for keyword in keywords)


def count_keywords(text, keywords):
    """
    Count how many keywords occur.
    """
    text = normalize(text)

    return sum(
        1 for keyword in keywords
        if keyword.lower() in text
    )


# ============================================================
# 5. OBVIOUS NOISE
# ============================================================

NOISE_PHRASES = {
    "first",
    "second",
    "third",
    "thanks",
    "thank you",
    "thank u",
    "tq",
    "tqtq",
    "hi",
    "hello",
    "lol",
    "lmao",
    "haha",
    "hahaha",
    "nice video",
    "great video",
    "good video",
    "love your video",
    "love this video",
    "new subscriber",
    "subscribed",
}


def is_noise(text):

    text = normalize(text)

    # Empty
    if text == "":
        return True

    # Remove @mentions
    no_mentions = re.sub(r"@\S+", "", text).strip()

    # If only emoji/punctuation/symbols remain
    if not re.search(r"[a-zA-Z0-9À-ÿ]", no_mentions):
        return True

    # Simplified text
    simple = re.sub(
        r"[^a-zA-Z0-9À-ÿ\s']",
        "",
        no_mentions
    )

    simple = re.sub(r"\s+", " ", simple).strip()

    if simple in NOISE_PHRASES:
        return True

    return False


df["is_noise"] = df["clean_text"].apply(is_noise)


# ============================================================
# 6. DETECT COMMENT TARGET
# ============================================================
#
# We want to distinguish:
#
# destination -> about Malaysia/place/travel experience
# creator     -> about YouTuber/video/editing/person
# unrelated   -> obviously unrelated
# unknown     -> not enough evidence


CREATOR_KEYWORDS = [
    "your video",
    "great video",
    "nice video",
    "good video",
    "amazing video",
    "love your videos",
    "love your vlog",
    "editing",
    "transition",
    "camera",
    "iphone",
    "subscriber",
    "subscribe",
    "channel",
    "content",
    "youtuber",
    "vlogger",
    "your videos",
    "next video",
    "new video",
    "beautiful video",
    "great job bro",
    "great job sis",
    "you guys",
    "your family",
]


TOURISM_GENERAL = [
    "travel",
    "travelling",
    "traveling",
    "traveler",
    "traveller",
    "tourist",
    "tourism",
    "trip",
    "vacation",
    "holiday",
    "visit",
    "visited",
    "visiting",
    "destination",
    "itinerary",
    "backpacking",
    "backpacker",
    "sightseeing",
]


MALAYSIA_TERMS = [
    "malaysia",
    "malaysian",
    "kl",
    "kuala lumpur",
    "penang",
    "georgetown",
    "george town",
    "langkawi",
    "malacca",
    "melaka",
    "johor",
    "johor bahru",
    "sabah",
    "sarawak",
    "borneo",
    "putrajaya",
    "ipoh",
    "perak",
    "selangor",
    "terengganu",
    "kelantan",
    "pahang",
    "kedah",
    "perlis",
    "negeri sembilan",
    "tioman",
    "perhentian",
    "redang",
    "semporna",
    "sipadan",
    "kinabalu",
    "kota kinabalu",
    "kuching",
    "genting",
    "cameron highlands",
]


TOURISM_EXPERIENCE_TERMS = [
    "hotel",
    "hostel",
    "resort",
    "beach",
    "island",
    "airport",
    "flight",
    "train",
    "bus",
    "grab",
    "taxi",
    "food",
    "restaurant",
    "street food",
    "hawker",
    "market",
    "museum",
    "temple",
    "mosque",
    "park",
    "waterfall",
    "mountain",
    "diving",
    "snorkeling",
    "scuba",
    "tour guide",
    "ticket",
    "entrance fee",
]


def identify_target(text):

    text = normalize(text)

    if is_noise(text):
        return "noise"

    creator_score = count_keywords(
        text,
        CREATOR_KEYWORDS
    )

    destination_score = (
        count_keywords(text, TOURISM_GENERAL)
        + count_keywords(text, MALAYSIA_TERMS)
        + count_keywords(text, TOURISM_EXPERIENCE_TERMS)
    )

    # Strong creator signal but no destination discussion
    if creator_score > 0 and destination_score == 0:
        return "creator"

    if destination_score > 0:
        return "destination"

    return "unknown"


df["target"] = df["clean_text"].apply(
    identify_target
)


# ============================================================
# 7. STAKEHOLDER IDENTIFICATION
# ============================================================
#
# IMPORTANT:
# Only classify LOCAL/VISITOR where the commenter explicitly
# provides evidence.
#
# Do NOT infer nationality from username, language, etc.


LOCAL_PATTERNS = [
    r"\bas a malaysian\b",
    r"\bi am malaysian\b",
    r"\bi'm malaysian\b",
    r"\bim malaysian\b",
    r"\bmalaysian here\b",
    r"\bas a local\b",
    r"\bi am a local\b",
    r"\bi'm a local\b",
    r"\blocal here\b",
    r"\bi live in malaysia\b",
    r"\bi live in kl\b",
    r"\bi live in kuala lumpur\b",
    r"\bborn and raised in malaysia\b",
    r"\bborn and raised in kl\b",
    r"\bfrom malaysia\b",
]


VISITOR_PATTERNS = [
    r"\bi visited malaysia\b",
    r"\bvisited malaysia\b",
    r"\bwhen i visited\b",
    r"\bi went to malaysia\b",
    r"\btravelled to malaysia\b",
    r"\btraveled to malaysia\b",
    r"\bmy trip to malaysia\b",
    r"\bmy malaysia trip\b",
    r"\bspent .* days in malaysia\b",
    r"\bspent .* days in kl\b",
    r"\bspent .* days in kuala lumpur\b",
    r"\bwhen i was in malaysia\b",
    r"\bwhen we were in malaysia\b",
    r"\bi stayed in malaysia\b",
    r"\bwe stayed in malaysia\b",
    r"\bi came to malaysia\b",
]


FUTURE_VISITOR_PATTERNS = [
    r"\bi want to visit\b",
    r"\bi want to go\b",
    r"\bplanning to visit\b",
    r"\bplanning a trip\b",
    r"\bwill visit\b",
    r"\bgoing to malaysia\b",
    r"\bvisit malaysia someday\b",
    r"\bmy bucket list\b",
]


def stakeholder(text):

    text = normalize(text)

    for pattern in LOCAL_PATTERNS:
        if re.search(pattern, text):
            return "local"

    for pattern in VISITOR_PATTERNS:
        if re.search(pattern, text):
            return "visitor"

    for pattern in FUTURE_VISITOR_PATTERNS:
        if re.search(pattern, text):
            return "potential_visitor"

    return "unknown"


df["stakeholder"] = df["clean_text"].apply(
    stakeholder
)


# ============================================================
# 8. TOURISM / SUSTAINABILITY ASPECTS
# ============================================================

ASPECT_KEYWORDS = {

    # ----------------------------
    # ENVIRONMENT
    # ----------------------------

    "cleanliness": [
        "clean",
        "dirty",
        "rubbish",
        "trash",
        "garbage",
        "litter",
        "waste",
        "plastic",
        "smelly",
        "polluted",
        "pollution",
    ],

    "beach_marine": [
        "beach",
        "sea",
        "ocean",
        "coast",
        "coastal",
        "coral",
        "reef",
        "marine",
        "diving",
        "snorkeling",
        "water quality",
    ],

    "nature_wildlife": [
        "nature",
        "forest",
        "jungle",
        "wildlife",
        "animal",
        "monkey",
        "turtle",
        "orangutan",
        "national park",
        "mountain",
        "waterfall",
        "biodiversity",
    ],

    "climate_weather": [
        "weather",
        "rain",
        "rainy",
        "monsoon",
        "flood",
        "flooding",
        "hot",
        "heat",
        "humid",
        "climate",
    ],


    # ----------------------------
    # SOCIAL
    # ----------------------------

    "crowding": [
        "crowded",
        "crowd",
        "too many tourists",
        "overtourism",
        "busy",
        "packed",
        "queue",
        "long queue",
        "too many people",
    ],

    "safety": [
        "safe",
        "unsafe",
        "dangerous",
        "crime",
        "scam",
        "scammer",
        "pickpocket",
        "robbery",
        "security",
    ],

    "hospitality": [
        "friendly",
        "helpful",
        "kind people",
        "nice people",
        "locals",
        "local people",
        "hospitality",
        "welcoming",
        "rude",
    ],

    "culture_heritage": [
        "culture",
        "cultural",
        "heritage",
        "traditional",
        "tradition",
        "temple",
        "mosque",
        "historic",
        "history",
        "local culture",
        "authentic",
    ],


    # ----------------------------
    # ECONOMIC
    # ----------------------------

    "price_value": [
        "cheap",
        "expensive",
        "price",
        "prices",
        "cost",
        "costly",
        "affordable",
        "overpriced",
        "value for money",
        "ringgit",
        "rm ",
    ],

    "local_business": [
        "local business",
        "small business",
        "hawker",
        "market",
        "vendor",
        "seller",
        "shop",
        "restaurant",
        "local restaurant",
    ],

    "food": [
        "food",
        "nasi lemak",
        "laksa",
        "satay",
        "char kway teow",
        "roti canai",
        "restaurant",
        "hawker",
        "street food",
        "delicious",
        "sedap",
        "makanan",
    ],


    # ----------------------------
    # INFRASTRUCTURE
    # ----------------------------

    "transport": [
        "public transport",
        "transport",
        "train",
        "lrt",
        "mrt",
        "monorail",
        "bus",
        "grab",
        "taxi",
        "airport",
        "flight",
        "traffic",
        "road",
        "parking",
    ],

    "accessibility": [
        "accessible",
        "accessibility",
        "wheelchair",
        "disabled",
        "elderly",
        "walking",
        "walkable",
        "pedestrian",
        "sidewalk",
    ],

    "facilities": [
        "toilet",
        "toilets",
        "restroom",
        "facility",
        "facilities",
        "signage",
        "sign",
        "wifi",
        "internet",
        "parking",
    ],

    "accommodation": [
        "hotel",
        "hostel",
        "resort",
        "airbnb",
        "accommodation",
        "room",
        "stay",
        "homestay",
    ],

    "seasonality": [
        "school holiday",
        "school holidays",
        "public holiday",
        "public holidays",
        "peak season",
        "off season",
        "high season",
        "low season",
        "monsoon season",
        "holiday season",
    ],
}


def detect_aspects(text):

    text = normalize(text)

    found = []

    for aspect, keywords in ASPECT_KEYWORDS.items():

        if contains_any(text, keywords):
            found.append(aspect)

    if not found:
        return "none"

    return ";".join(found)


df["tourism_aspect"] = df["clean_text"].apply(
    detect_aspects
)


# ============================================================
# 9. MAP ASPECTS -> SUSTAINABILITY DIMENSIONS
# ============================================================

DIMENSION_MAP = {

    "cleanliness": "environmental",
    "beach_marine": "environmental",
    "nature_wildlife": "environmental",
    "climate_weather": "environmental",

    "crowding": "social",
    "safety": "social",
    "hospitality": "social",
    "culture_heritage": "social",

    "price_value": "economic",
    "local_business": "economic",
    "food": "economic",

    "transport": "infrastructure",
    "accessibility": "infrastructure",
    "facilities": "infrastructure",
    "accommodation": "infrastructure",
    "seasonality": "infrastructure",
}


def get_dimensions(aspects):

    if aspects == "none":
        return "none"

    dimensions = set()

    for aspect in aspects.split(";"):

        if aspect in DIMENSION_MAP:
            dimensions.add(
                DIMENSION_MAP[aspect]
            )

    if not dimensions:
        return "none"

    return ";".join(sorted(dimensions))


df["sustainability_dimension"] = (
    df["tourism_aspect"]
    .apply(get_dimensions)
)


# ============================================================
# 10. DESTINATION EXTRACTION
# ============================================================

DESTINATIONS = {

    "Kuala Lumpur": [
        "kuala lumpur",
        " kl ",
        " kl.",
    ],

    "Penang": [
        "penang",
        "georgetown",
        "george town",
    ],

    "Langkawi": [
        "langkawi",
    ],

    "Melaka": [
        "melaka",
        "malacca",
    ],

    "Johor": [
        "johor",
        "johor bahru",
        "jb",
    ],

    "Sabah": [
        "sabah",
        "kota kinabalu",
        "kinabalu",
        "semporna",
        "sipadan",
    ],

    "Sarawak": [
        "sarawak",
        "kuching",
    ],

    "Putrajaya": [
        "putrajaya",
    ],

    "Perak": [
        "perak",
        "ipoh",
    ],

    "Pahang": [
        "pahang",
        "cameron highlands",
        "genting",
    ],

    "Terengganu": [
        "terengganu",
        "redang",
        "perhentian",
    ],

    "Malaysia": [
        "malaysia",
        "malaysian",
    ],
}


def detect_destination(text):

    text = " " + normalize(text) + " "

    found = []

    for destination, terms in DESTINATIONS.items():

        if any(term in text for term in terms):
            found.append(destination)

    # If specific state/place found, Malaysia doesn't need
    # to appear separately
    if len(found) > 1 and "Malaysia" in found:
        found.remove("Malaysia")

    if not found:
        return "unknown"

    return ";".join(found)


df["destination"] = df["clean_text"].apply(
    detect_destination
)


# ============================================================
# 11. SIMPLE RULE-BASED SENTIMENT
# ============================================================
#
# This is ONLY a first-pass sentiment system.
# Later you can replace this with a multilingual transformer.


POSITIVE_WORDS = [
    "good",
    "great",
    "amazing",
    "beautiful",
    "awesome",
    "excellent",
    "love",
    "loved",
    "like",
    "best",
    "perfect",
    "friendly",
    "helpful",
    "clean",
    "safe",
    "cheap",
    "affordable",
    "delicious",
    "nice",
    "wonderful",
    "fantastic",
    "enjoy",
    "enjoyed",
    "recommend",
    "recommended",
    "worth",
    "underrated",
    "cantik",
    "cantek",
    "sedap",
]


NEGATIVE_WORDS = [
    "bad",
    "terrible",
    "awful",
    "hate",
    "worst",
    "dirty",
    "unsafe",
    "dangerous",
    "expensive",
    "overpriced",
    "crowded",
    "polluted",
    "pollution",
    "rubbish",
    "garbage",
    "trash",
    "scam",
    "scammer",
    "disappointed",
    "disappointing",
    "boring",
    "poor",
    "problem",
    "traffic",
]


def simple_sentiment(text):

    text = normalize(text)

    positive = count_keywords(
        text,
        POSITIVE_WORDS
    )

    negative = count_keywords(
        text,
        NEGATIVE_WORDS
    )

    if positive > negative:
        return "positive"

    elif negative > positive:
        return "negative"

    else:
        return "neutral"


df["sentiment"] = df["clean_text"].apply(
    simple_sentiment
)


# ============================================================
# 12. HELPFULNESS SCORE 0 - 5
# ============================================================
#
# 0 = useless/noise
# 1 = very low information
# 2 = general tourism opinion
# 3 = specific tourism information
# 4 = useful sustainability observation
# 5 = highly informative / actionable


EXPERIENCE_PATTERNS = [
    r"\bi visited\b",
    r"\bwe visited\b",
    r"\bi went\b",
    r"\bwe went\b",
    r"\bi stayed\b",
    r"\bwe stayed\b",
    r"\bi travelled\b",
    r"\bi traveled\b",
    r"\bwhen i was\b",
    r"\bwhen we were\b",
    r"\bspent .* day",
    r"\bmy experience\b",
    r"\bin my experience\b",
]


ACTIONABLE_WORDS = [
    "should",
    "need to",
    "needs to",
    "must",
    "recommend",
    "avoid",
    "better",
    "improve",
    "problem",
    "issue",
    "too many",
    "too much",
    "not enough",
    "please",
    "tip",
    "advice",
]


def helpfulness_score(row):

    text = normalize(row["clean_text"])

    # ------------------------
    # Score 0
    # ------------------------

    if row["is_noise"]:
        return 0

    if row["target"] == "creator":
        return 0

    score = 1


    # ------------------------
    # Tourism context
    # ------------------------

    if row["target"] == "destination":
        score += 1


    # ------------------------
    # Contains a tourism aspect
    # ------------------------

    if row["tourism_aspect"] != "none":
        score += 1


    # ------------------------
    # First-hand experience
    # ------------------------

    if any(
        re.search(pattern, text)
        for pattern in EXPERIENCE_PATTERNS
    ):
        score += 1


    # ------------------------
    # Specific actionable information
    # ------------------------

    if contains_any(
        text,
        ACTIONABLE_WORDS
    ):
        score += 1


    # ------------------------
    # Detailed sustainability comment
    # ------------------------

    if (
        row["sustainability_dimension"] != "none"
        and row["word_count_new"] >= 10
    ):
        score += 1


    # ------------------------
    # Very short comments cannot be
    # considered highly informative
    # ------------------------

    if row["word_count_new"] <= 2:
        score = min(score, 2)


    # Max 5
    return min(score, 5)


df["helpfulness_score"] = df.apply(
    helpfulness_score,
    axis=1
)


# ============================================================
# 13. CREATE ANALYSIS FLAGS
# ============================================================

# Broad tourism comments
df["tourism_useful"] = (
    (df["target"] == "destination")
    &
    (df["helpfulness_score"] >= 2)
)


# Strong comments for analysis
df["analysis_ready"] = (
    (df["target"] == "destination")
    &
    (df["sustainability_dimension"] != "none")
    &
    (df["helpfulness_score"] >= 3)
)


# Very strong / actionable comments
df["high_value"] = (
    df["analysis_ready"]
    &
    (df["helpfulness_score"] >= 4)
)


# ============================================================
# 14. WEIGHT USING LIKES
# ============================================================
#
# Do NOT let one viral comment dominate everything.
# log1p compresses large like counts.

df["like_count"] = pd.to_numeric(
    df["like_count"],
    errors="coerce"
).fillna(0)

df["engagement_weight"] = np.log1p(
    df["like_count"]
)


# ============================================================
# 15. EXPORT DATASETS
# ============================================================

# Complete classified dataset
df.to_csv(
    "comments_classified.csv",
    index=False
)


# Tourism-related useful comments
tourism_df = df[
    df["tourism_useful"]
].copy()

tourism_df.to_csv(
    "tourism_useful_comments.csv",
    index=False
)


# Sustainability dataset
sustainable_df = df[
    df["analysis_ready"]
].copy()

sustainable_df.to_csv(
    "sustainable_tourism_comments.csv",
    index=False
)


# Highest-value comments
high_value_df = df[
    df["high_value"]
].copy()

high_value_df.to_csv(
    "high_value_comments.csv",
    index=False
)


# Malaysian/local comments
local_df = sustainable_df[
    sustainable_df["stakeholder"] == "local"
].copy()

local_df.to_csv(
    "local_sustainability_comments.csv",
    index=False
)


# Visitor comments
visitor_df = sustainable_df[
    sustainable_df["stakeholder"] == "visitor"
].copy()

visitor_df.to_csv(
    "visitor_sustainability_comments.csv",
    index=False
)


# ============================================================
# 16. SUMMARY
# ============================================================

print("\n" + "=" * 70)
print("PROCESSING COMPLETE")
print("=" * 70)

print("\nTotal comments:")
print(len(df))

print("\nTARGET:")
print(df["target"].value_counts())

print("\nSTAKEHOLDER:")
print(df["stakeholder"].value_counts())

print("\nHELPFULNESS:")
print(
    df["helpfulness_score"]
    .value_counts()
    .sort_index()
)

print("\nSUSTAINABILITY DIMENSION:")
print(
    df["sustainability_dimension"]
    .value_counts()
    .head(20)
)

print("\nSENTIMENT:")
print(
    sustainable_df["sentiment"]
    .value_counts()
)

print("\nTourism useful:")
print(len(tourism_df))

print("\nSustainability analysis ready:")
print(len(sustainable_df))

print("\nHigh-value comments:")
print(len(high_value_df))

print("\nExplicit local comments:")
print(len(local_df))

print("\nExplicit visitor comments:")
print(len(visitor_df))


# ============================================================
# 17. SHOW HIGH-VALUE EXAMPLES
# ============================================================

columns_to_show = [
    "clean_text",
    "stakeholder",
    "destination",
    "tourism_aspect",
    "sustainability_dimension",
    "sentiment",
    "helpfulness_score",
    "like_count"
]

print("\n" + "=" * 70)
print("HIGH-VALUE EXAMPLES")
print("=" * 70)

print(
    high_value_df[
        columns_to_show
    ]
    .sort_values(
        ["helpfulness_score", "like_count"],
        ascending=False
    )
    .head(30)
    .to_string(index=False)
)
```

## Why this extends rather than replaces `sentiment`

The existing `sentiment` column (keyword-based: more positive words than negative words → positive) is a
reasonable first pass, but two things limit it: it can't handle mixed sentiment in one comment (*"food is good, but
country is fucking suck"* currently scores **positive**, purely because "good" is present), and it doesn't
generalize to the ~8% of `analysis_ready` comments that aren't in English. The plan below keeps `sentiment` intact
as a labelled baseline (renamed `rule_sentiment`), adds a multilingual transformer baseline that needs no manual
labelling, and gates a fine-tuned tourism-domain model behind actually having gold labels to fine-tune on —
each stage is checked against this data rather than assumed to work.

## Stage 1 — Prepare the sentiment-ready dataset

Scope to `analysis_ready == True` (the 4,755 comments that are on-topic, tourism-relevant, and have a
sustainability dimension attached — already filtered upstream). Preserve the existing rule-based label under a
new name so later stages can be compared against it rather than overwriting it.

```python
import pandas as pd
import numpy as np

df = pd.read_csv("final.csv")

sentiment_df = df[df["analysis_ready"] == True].copy().reset_index(drop=True)
sentiment_df = sentiment_df.rename(columns={"sentiment": "rule_sentiment"})

print("analysis_ready comments:", len(sentiment_df))
print("\nrule_sentiment baseline (treat as OLD baseline, not ground truth):")
counts = sentiment_df["rule_sentiment"].value_counts()
pct = (counts / counts.sum() * 100).round(1)
print(pd.DataFrame({"count": counts, "pct": pct}))
```

```python
# Recompute language detection on the WORKING SET ONLY (13s vs minutes on all 23k rows) -
# the version in cells 23-25 above was silently dropped when cell 26 reloaded df from v1.csv.
try:
    from lingua import LanguageDetectorBuilder
except ImportError:
    import sys, subprocess
    subprocess.check_call([sys.executable, "-m", "pip", "install", "lingua-language-detector", "-q"])
    from lingua import LanguageDetectorBuilder

detector = LanguageDetectorBuilder.from_all_languages().build()

def detect_language(text):
    try:
        result = detector.detect_language_of(str(text))
        return result.iso_code_639_1.name if result else "unknown"
    except Exception:
        return "unknown"

sentiment_df["language"] = sentiment_df["clean_text"].apply(detect_language)

# Coarse EN / non-EN routing flag. Fine-grained codes on very short comments are noisy
# (lingua misfires on a handful of words) - what actually matters for TourBERT-vs-XLM-R
# routing is just "is this English or not".
sentiment_df["is_english"] = sentiment_df["language"].eq("EN")

print(sentiment_df["language"].value_counts().head(10))
print(f"\nEnglish share: {sentiment_df['is_english'].mean():.1%} "
      f"({sentiment_df['is_english'].sum()} of {len(sentiment_df)})")
```

## Stage 2 — Multilingual pretrained sentiment baseline (XLM-R)

**Requires internet access to huggingface.co and, ideally, a GPU — run this in Colab, not in an offline/sandboxed
environment.** `cardiffnlp/twitter-xlm-roberta-base-sentiment` needs no fine-tuning and covers every language in
this dataset in one pass, so it becomes the baseline every other model is compared against. The result is cached
to `sentiment_xlmr.csv` so re-running the notebook doesn't re-download/re-infer unnecessarily.

```python
import os

XLMR_CACHE = "sentiment_xlmr.csv"
MODEL_NAME = "cardiffnlp/twitter-xlm-roberta-base-sentiment"

if os.path.exists(XLMR_CACHE):
    print(f"Loading cached XLM-R predictions from {XLMR_CACHE} (delete this file to force re-inference).")
    xlmr_predictions = pd.read_csv(XLMR_CACHE)
    assert len(xlmr_predictions) == len(sentiment_df), (
        "Cached predictions don't match the current sentiment_df length - delete the cache and re-run."
    )
else:
    import torch
    from transformers import AutoTokenizer, AutoModelForSequenceClassification
    from tqdm.auto import tqdm

    tokenizer = AutoTokenizer.from_pretrained(MODEL_NAME)
    model = AutoModelForSequenceClassification.from_pretrained(MODEL_NAME)

    device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
    model.to(device)
    model.eval()
    print("Using:", device, "- inference will be slow on CPU for 4,755 comments; a Colab GPU runtime is recommended.")

    def predict_sentiment_batch(texts, batch_size=32):
        results = []
        label_names = ["negative", "neutral", "positive"]
        for i in tqdm(range(0, len(texts), batch_size)):
            batch = texts[i:i + batch_size]
            encoded = tokenizer(batch, padding=True, truncation=True, max_length=128, return_tensors="pt")
            encoded = {k: v.to(device) for k, v in encoded.items()}
            with torch.no_grad():
                output = model(**encoded)
            probs = torch.softmax(output.logits, dim=1).cpu().numpy()
            for p in probs:
                results.append({
                    "model_sentiment": label_names[p.argmax()],
                    "sentiment_confidence": float(p.max()),
                    "negative_probability": float(p[0]),
                    "neutral_probability": float(p[1]),
                    "positive_probability": float(p[2]),
                })
        return pd.DataFrame(results)

    texts = sentiment_df["clean_text"].fillna("").astype(str).tolist()
    xlmr_predictions = predict_sentiment_batch(texts)
    xlmr_predictions.to_csv(XLMR_CACHE, index=False)
    print(f"Saved predictions to {XLMR_CACHE}")

xlmr_predictions.index = sentiment_df.index
sentiment_df = pd.concat(
    [sentiment_df.drop(columns=xlmr_predictions.columns, errors="ignore"), xlmr_predictions],
    axis=1
)
sentiment_df[["clean_text", "rule_sentiment", "model_sentiment", "sentiment_confidence"]].head(10)
```

**Why probabilities are stored, not just the label:** *"Beach is dirty"* at `negative_probability=0.91` and *"Food
is okay, but nothing amazing"* at `neutral_probability=0.71`/`positive_probability=0.17` both currently collapse to
one word each under the old rule system. Keeping all three probabilities plus `sentiment_confidence` lets the
dashboard distinguish **strong** sentiment from **uncertain** sentiment — e.g. only surfacing a comment under "Top
Negative Issues" when `negative_probability` clears a threshold, not just when it happens to win a 3-way tie.

```python
# Sanity check with NO gold labels needed yet: how often do the rule-based baseline
# and the XLM-R baseline agree? Large disagreement is expected and informative, not a bug -
# it's exactly the mixed-sentiment / non-English cases the rule system couldn't handle.
agreement = (sentiment_df["rule_sentiment"] == sentiment_df["model_sentiment"]).mean()
print(f"Rule-based vs XLM-R agreement: {agreement:.1%}\n")
print("Confusion matrix (rows = rule_sentiment, columns = model_sentiment):")
print(pd.crosstab(sentiment_df["rule_sentiment"], sentiment_df["model_sentiment"]))
```

## Stage 3 — Build a gold-standard annotation sample (stratified, not naive random)

A plain random sample of 1,000 comments would inherit the rule-based class imbalance almost exactly
(~49% positive / ~43% neutral / ~7% negative, per Stage 1), leaving only ~72 negative examples to evaluate against
— too few to trust a negative-class F1 score. Since negative comments are exactly the sustainability problems this
project needs to detect, the sample below **keeps every rule-flagged negative comment** and fills the rest evenly
from positive/neutral, so all three classes have a comparable number of examples to hand-label and evaluate on.

```python
TARGET_N = 1000

neg_all = sentiment_df[sentiment_df["rule_sentiment"] == "negative"]
remaining_budget = max(TARGET_N - len(neg_all), 0)
per_class_remaining = remaining_budget // 2

pos_available = sentiment_df[sentiment_df["rule_sentiment"] == "positive"]
neu_available = sentiment_df[sentiment_df["rule_sentiment"] == "neutral"]

pos_sample = pos_available.sample(n=min(per_class_remaining, len(pos_available)), random_state=42)
neu_sample = neu_available.sample(n=min(per_class_remaining, len(neu_available)), random_state=42)

annotation_sample = (
    pd.concat([neg_all, pos_sample, neu_sample])
    .sample(frac=1, random_state=42)   # shuffle so labelling order doesn't track rule_sentiment
    .reset_index(drop=True)
)
annotation_sample["gold_sentiment"] = ""

print("Annotation sample composition (target: roughly even across classes):")
print(annotation_sample["rule_sentiment"].value_counts())

export_cols = ["clean_text", "tourism_aspect", "sustainability_dimension", "rule_sentiment", "gold_sentiment"]
# include the model baseline as a cross-check column for the annotator, if Stage 2 has been run
if "model_sentiment" in annotation_sample.columns:
    export_cols.insert(-1, "model_sentiment")

annotation_sample[export_cols].to_csv("sentiment_annotation.csv", index=False)
print("\nSaved sentiment_annotation.csv - open it and fill 'gold_sentiment' with positive/neutral/negative for each row.")
```

## Stage 4 — Fine-tune TourBERT (gated — do not "Run All" through this until labelling is done)

TourBERT (`veroman/TourBERT`) is a tourism-domain BERT **encoder** — it has no ready-made sentiment head, so it
must be fine-tuned on the gold labels from Stage 3 before it can be compared to anything. This is set behind an
explicit flag rather than a `try/except` around a missing file, so a top-to-bottom "Run All" never silently skips
training when labels *are* ready, and never crashes when they aren't:

1. Open `sentiment_annotation.csv`, fill in `gold_sentiment` for as many rows as you can (aim for 800–1,500;
   the stratified sample above already guarantees a reasonable number of each class).
2. Save the file, then flip `RUN_TOURBERT_TRAINING = True` below.

**Also requires internet + GPU (Colab), same as Stage 2.**

```python
RUN_TOURBERT_TRAINING = False  # flip to True once sentiment_annotation.csv has gold_sentiment filled in

MIN_LABELLED_ROWS = 100  # below this, a fine-tuned model isn't meaningfully evaluable

if RUN_TOURBERT_TRAINING:
    labels = {"negative": 0, "neutral": 1, "positive": 2}

    labeled_df = pd.read_csv("sentiment_annotation.csv")
    labeled_df = labeled_df[labeled_df["gold_sentiment"].isin(labels.keys())].copy()

    if len(labeled_df) < MIN_LABELLED_ROWS:
        print(f"Only {len(labeled_df)} labelled rows found (need >= {MIN_LABELLED_ROWS}). "
              f"Label more of sentiment_annotation.csv before training - skipping for now.")
        RUN_TOURBERT_TRAINING = False
    else:
        labeled_df["label"] = labeled_df["gold_sentiment"].map(labels)
        print(labeled_df["gold_sentiment"].value_counts())
else:
    print("RUN_TOURBERT_TRAINING is False - skipping fine-tuning. "
          "This is expected until sentiment_annotation.csv has been hand-labelled.")
```

```python
if RUN_TOURBERT_TRAINING:
    from sklearn.model_selection import train_test_split
    from datasets import Dataset
    from transformers import AutoTokenizer, AutoModelForSequenceClassification, TrainingArguments, Trainer
    import numpy as np
    from sklearn.metrics import accuracy_score, precision_recall_fscore_support

    train_df, temp_df = train_test_split(
        labeled_df, test_size=0.30, stratify=labeled_df["label"], random_state=42
    )
    validation_df, test_df = train_test_split(
        temp_df, test_size=0.50, stratify=temp_df["label"], random_state=42
    )
    print(f"Train={len(train_df)}, Validation={len(validation_df)}, Test={len(test_df)}")

    TOURBERT = "veroman/TourBERT"
    tokenizer = AutoTokenizer.from_pretrained(TOURBERT)
    model = AutoModelForSequenceClassification.from_pretrained(
        TOURBERT, num_labels=3,
        id2label={0: "negative", 1: "neutral", 2: "positive"},
        label2id=labels,
    )
    # NOTE: a "classification layer newly initialized" warning here is expected -
    # TourBERT knows tourism language, the sentiment head is new and needs this fine-tuning.

    def tokenize(batch):
        return tokenizer(batch["clean_text"], truncation=True, padding="max_length", max_length=128)

    train_dataset = Dataset.from_pandas(train_df[["clean_text", "label"]]).map(tokenize, batched=True)
    validation_dataset = Dataset.from_pandas(validation_df[["clean_text", "label"]]).map(tokenize, batched=True)
    test_dataset = Dataset.from_pandas(test_df[["clean_text", "label"]]).map(tokenize, batched=True)

    def compute_metrics(pred):
        labels_true = pred.label_ids
        preds = np.argmax(pred.predictions, axis=1)
        precision, recall, f1, _ = precision_recall_fscore_support(
            labels_true, preds, average="macro", zero_division=0
        )
        return {"accuracy": accuracy_score(labels_true, preds),
                "precision_macro": precision, "recall_macro": recall, "f1_macro": f1}

    training_args = TrainingArguments(
        output_dir="./tourbert_sentiment",
        learning_rate=2e-5,
        per_device_train_batch_size=16,
        per_device_eval_batch_size=32,
        num_train_epochs=4,
        weight_decay=0.01,
        eval_strategy="epoch",
        save_strategy="epoch",
        load_best_model_at_end=True,
        metric_for_best_model="f1_macro",
        greater_is_better=True,
    )
    trainer = Trainer(
        model=model, args=training_args,
        train_dataset=train_dataset, eval_dataset=validation_dataset,
        tokenizer=tokenizer, compute_metrics=compute_metrics,
    )
    trainer.train()

    tourbert_test_results = trainer.evaluate(test_dataset)
    print(tourbert_test_results)

    # Apply the fine-tuned model to the full analysis-ready set (not just the labelled sample)
    full_dataset = Dataset.from_pandas(sentiment_df[["clean_text"]]).map(tokenize, batched=True)
    tourbert_logits = trainer.predict(full_dataset).predictions
    tourbert_probs = torch.softmax(torch.tensor(tourbert_logits), dim=1).numpy()
    id2label = {0: "negative", 1: "neutral", 2: "positive"}
    sentiment_df["tourbert_sentiment"] = [id2label[p.argmax()] for p in tourbert_probs]
    sentiment_df["tourbert_confidence"] = tourbert_probs.max(axis=1)
    sentiment_df.to_csv("sentiment_with_tourbert.csv", index=False)
    print("Saved sentiment_with_tourbert.csv")
```

## Stage 5 — Compare rule-based vs XLM-R vs fine-tuned TourBERT

**Macro F1, not accuracy, is the metric that matters here.** With a class split like 70% positive / 25% neutral /
5% negative, a model can post a deceptively high accuracy by mostly predicting "positive" — Macro F1 forces
reasonable performance on all three classes, including negative, which is exactly the class sustainability
problems show up in. Nothing is assumed in advance: if XLM-R matches or beats TourBERT on this data, use XLM-R.

```python
if RUN_TOURBERT_TRAINING and "tourbert_test_results" in dir():
    from sklearn.metrics import accuracy_score, precision_recall_fscore_support

    def score_against_gold(gold_labels, pred_labels):
        precision, recall, f1, _ = precision_recall_fscore_support(
            gold_labels, pred_labels, average="macro", zero_division=0, labels=["negative","neutral","positive"]
        )
        _, neg_recall, _, _ = precision_recall_fscore_support(
            gold_labels, pred_labels, average=None, zero_division=0, labels=["negative","neutral","positive"]
        )[0:3]
        return {
            "accuracy": accuracy_score(gold_labels, pred_labels),
            "macro_f1": f1,
            "negative_recall": recall[0] if hasattr(recall, "__len__") else None,
        }

    gold_test = test_df["gold_sentiment"]
    comparison_rows = []
    for name, col in [("Keyword baseline", "rule_sentiment"), ("XLM-R", "model_sentiment")]:
        preds = sentiment_df.loc[test_df.index, col]
        comparison_rows.append({"model": name, **score_against_gold(gold_test, preds)})
    comparison_rows.append({
        "model": "Fine-tuned TourBERT",
        "accuracy": tourbert_test_results.get("eval_accuracy"),
        "macro_f1": tourbert_test_results.get("eval_f1_macro"),
        "negative_recall": None,  # per-class recall available via sklearn.metrics.classification_report if needed
    })
    comparison_table = pd.DataFrame(comparison_rows).round(3)
    print(comparison_table.to_string(index=False))
    print("\nDon't assume TourBERT wins - pick whichever row has the highest macro_f1 for FINAL_SENTIMENT_SOURCE below.")
else:
    print("Skipped - requires RUN_TOURBERT_TRAINING=True and a completed Stage 4 run "
          "(needs gold_sentiment labels in sentiment_annotation.csv).")
```

## Stage 6 — Choose the final sentiment source (auto-detects what's actually available)

Rather than hardcoding a choice, this picks the best available model in order of sophistication — fine-tuned
TourBERT if Stage 4/5 showed it's the strongest, else the XLM-R baseline, else the original rule-based labels — so
the same notebook works correctly whichever stages have actually been run.

```python
if "tourbert_sentiment" in sentiment_df.columns:
    FINAL_SENTIMENT_SOURCE = "tourbert_sentiment"
    FINAL_CONFIDENCE_SOURCE = "tourbert_confidence"
elif "model_sentiment" in sentiment_df.columns:
    FINAL_SENTIMENT_SOURCE = "model_sentiment"
    FINAL_CONFIDENCE_SOURCE = "sentiment_confidence"
else:
    FINAL_SENTIMENT_SOURCE = "rule_sentiment"
    FINAL_CONFIDENCE_SOURCE = None

sentiment_df["final_sentiment"] = sentiment_df[FINAL_SENTIMENT_SOURCE]
sentiment_df["final_confidence"] = sentiment_df[FINAL_CONFIDENCE_SOURCE] if FINAL_CONFIDENCE_SOURCE else np.nan

print(f"Using '{FINAL_SENTIMENT_SOURCE}' as final_sentiment for all dashboard aggregation below.")
print(sentiment_df["final_sentiment"].value_counts())
```

## Stage 7 — Dashboard-ready aggregation exports

Turns the 4,755 scored comments into the exact aggregated shapes a dashboard needs — never raw comments. Every
aggregation reports **both** a simple count-based `net_sentiment` and an **engagement-weighted** version (using the
`engagement_weight` = `log1p(like_count)` column that the original pipeline already computed but never used) so one
highly-liked comment doesn't get the same influence as one with zero likes, without letting it completely dominate
either.

$$\text{net\_sentiment} = \%positive - \%negative \quad (-100 = \text{all negative}, \; 0 = \text{balanced}, \; +100 = \text{all positive})$$

```python
import os
os.makedirs("dashboard_exports", exist_ok=True)

SENT_COL = "final_sentiment"

def net_sentiment(s):
    c = s.value_counts(normalize=True) * 100
    return round(c.get("positive", 0.0) - c.get("negative", 0.0), 1)

def weighted_net_sentiment(labels, weights):
    pos_w = weights[labels == "positive"].sum()
    neg_w = weights[labels == "negative"].sum()
    tot_w = weights.sum()
    return round((pos_w - neg_w) / tot_w * 100, 1) if tot_w > 0 else np.nan
```

```python
# ---- 7a. Overall headline KPI ----
overall_counts = sentiment_df[SENT_COL].value_counts()
overall_pct = (overall_counts / overall_counts.sum() * 100).round(1)

overall_summary = pd.DataFrame({
    "metric": ["analyzed_comments", "positive_pct", "neutral_pct", "negative_pct",
               "net_sentiment", "weighted_net_sentiment"],
    "value": [
        len(sentiment_df),
        overall_pct.get("positive", 0.0),
        overall_pct.get("neutral", 0.0),
        overall_pct.get("negative", 0.0),
        net_sentiment(sentiment_df[SENT_COL]),
        weighted_net_sentiment(sentiment_df[SENT_COL], sentiment_df["engagement_weight"]),
    ]
})
print(overall_summary.to_string(index=False))
overall_summary.to_csv("dashboard_exports/agg_overall.csv", index=False)
```

```python
# ---- 7b. Sustainability dimension ----
dim_df = sentiment_df.assign(
    sustainability_dimension=sentiment_df["sustainability_dimension"].str.split(";")
).explode("sustainability_dimension")
dim_df = dim_df[dim_df["sustainability_dimension"] != "none"]

dim_agg = dim_df.groupby("sustainability_dimension").apply(
    lambda g: pd.Series({
        "n": len(g),
        "positive_pct": round((g[SENT_COL] == "positive").mean() * 100, 1),
        "neutral_pct": round((g[SENT_COL] == "neutral").mean() * 100, 1),
        "negative_pct": round((g[SENT_COL] == "negative").mean() * 100, 1),
        "net_sentiment": net_sentiment(g[SENT_COL]),
        "weighted_net_sentiment": weighted_net_sentiment(g[SENT_COL], g["engagement_weight"]),
    })
).reset_index().sort_values("net_sentiment")

print(dim_agg.to_string(index=False))
dim_agg.to_csv("dashboard_exports/agg_dimension.csv", index=False)
```

```python
# ---- 7c. Tourism aspect (powers a "Top Negative Issues" ranking) ----
MIN_N_FOR_RANKING = 20  # hide aspects with too few comments to trust a percentage

asp_df = sentiment_df.assign(
    tourism_aspect=sentiment_df["tourism_aspect"].str.split(";")
).explode("tourism_aspect")
asp_df = asp_df[asp_df["tourism_aspect"] != "none"]

asp_agg = asp_df.groupby("tourism_aspect").apply(
    lambda g: pd.Series({
        "n": len(g),
        "positive_pct": round((g[SENT_COL] == "positive").mean() * 100, 1),
        "negative_pct": round((g[SENT_COL] == "negative").mean() * 100, 1),
        "net_sentiment": net_sentiment(g[SENT_COL]),
        "weighted_net_sentiment": weighted_net_sentiment(g[SENT_COL], g["engagement_weight"]),
    })
).reset_index()
asp_agg["low_sample_warning"] = asp_agg["n"] < MIN_N_FOR_RANKING

print("Top negative-leaning aspects (min n=%d):" % MIN_N_FOR_RANKING)
print(asp_agg[~asp_agg["low_sample_warning"]].sort_values("negative_pct", ascending=False).head(8).to_string(index=False))
asp_agg.to_csv("dashboard_exports/agg_aspect.csv", index=False)
```

```python
# ---- 7d. Destination x dimension heatmap ----
MIN_N_FOR_DESTINATION = 20

dd_df = sentiment_df.assign(
    destination=sentiment_df["destination"].str.split(";"),
    sustainability_dimension=sentiment_df["sustainability_dimension"].str.split(";"),
).explode("destination").explode("sustainability_dimension")
dd_df = dd_df[(dd_df["destination"] != "unknown") & (dd_df["sustainability_dimension"] != "none")]

dd_agg = dd_df.groupby(["destination", "sustainability_dimension"]).apply(
    lambda g: pd.Series({"n": len(g), "net_sentiment": net_sentiment(g[SENT_COL])})
).reset_index()

dest_totals = dd_df.groupby("destination").size()
keep_destinations = dest_totals[dest_totals >= MIN_N_FOR_DESTINATION].index
dd_agg_filtered = dd_agg[dd_agg["destination"].isin(keep_destinations)]

heatmap = dd_agg_filtered.pivot(index="destination", columns="sustainability_dimension", values="net_sentiment")
heatmap_n = dd_agg_filtered.pivot(index="destination", columns="sustainability_dimension", values="n")

print(f"Destination x dimension net-sentiment heatmap ({len(keep_destinations)} destinations with n>={MIN_N_FOR_DESTINATION}):")
print(heatmap.round(1))
dd_agg.to_csv("dashboard_exports/agg_destination_dimension.csv", index=False)
```

```python
# ---- 7e. Stakeholder (local / visitor / potential_visitor / unknown) ----
LOW_SAMPLE_THRESHOLD = 30

stake_agg = sentiment_df.groupby("stakeholder").apply(
    lambda g: pd.Series({
        "n": len(g),
        "positive_pct": round((g[SENT_COL] == "positive").mean() * 100, 1),
        "neutral_pct": round((g[SENT_COL] == "neutral").mean() * 100, 1),
        "negative_pct": round((g[SENT_COL] == "negative").mean() * 100, 1),
        "net_sentiment": net_sentiment(g[SENT_COL]),
    })
).reset_index()
# Machine-readable flag (not just a note a developer has to remember) so the dashboard
# can auto-render an "n=X, interpret with caution" caveat next to small groups.
stake_agg["low_sample_warning"] = stake_agg["n"] < LOW_SAMPLE_THRESHOLD

print(stake_agg.to_string(index=False))
stake_agg.to_csv("dashboard_exports/agg_stakeholder.csv", index=False)
```

```python
# ---- 7f. Sentiment over time ----
# NOTE: 2026 dates are NOT excluded here (see the Data Quality Note at the top) - if you
# decide to apply that filter upstream, this will automatically reflect it on next run.
ts_df = sentiment_df.copy()
ts_df["comment_published_at"] = pd.to_datetime(ts_df["comment_published_at"], errors="coerce", utc=True).dt.tz_localize(None)
ts_df["year_month"] = ts_df["comment_published_at"].dt.to_period("M").astype(str)
ts_df = ts_df.assign(
    sustainability_dimension=ts_df["sustainability_dimension"].str.split(";")
).explode("sustainability_dimension")
ts_df = ts_df[ts_df["sustainability_dimension"] != "none"]

ts_agg = ts_df.groupby(["year_month", "sustainability_dimension"]).apply(
    lambda g: pd.Series({"n": len(g), "net_sentiment": net_sentiment(g[SENT_COL])})
).reset_index()
ts_agg["low_sample_warning"] = ts_agg["n"] < 10

sparse_share = ts_agg["low_sample_warning"].mean()
print(f"{sparse_share:.0%} of year_month x dimension buckets have n<10 - monthly granularity is sparse. "
      f"Consider aggregating to quarters in the dashboard for a more stable trend line; "
      f"n and low_sample_warning are exported so the frontend can decide.")
ts_agg.to_csv("dashboard_exports/agg_timeseries_monthly.csv", index=False)

# Quarterly rollup, exported alongside monthly for a steadier default view
ts_df["year_quarter"] = ts_df["comment_published_at"].dt.to_period("Q").astype(str)
tsq_agg = ts_df.groupby(["year_quarter", "sustainability_dimension"]).apply(
    lambda g: pd.Series({"n": len(g), "net_sentiment": net_sentiment(g[SENT_COL])})
).reset_index()
tsq_agg.to_csv("dashboard_exports/agg_timeseries_quarterly.csv", index=False)
print(tsq_agg.tail(8).to_string(index=False))
```

```python
# ---- Also export the full scored comment-level dataset (for drill-down / search in the dashboard) ----
drilldown_cols = [
    "comment_id", "video_id", "clean_text", "comment_published_at", "like_count",
    "destination", "stakeholder", "tourism_aspect", "sustainability_dimension",
    "rule_sentiment", SENT_COL, "final_confidence", "helpfulness_score", "engagement_weight",
]
drilldown_cols = [c for c in drilldown_cols if c in sentiment_df.columns]
sentiment_df[drilldown_cols].to_csv("dashboard_exports/comments_scored.csv", index=False)

print("All dashboard exports written to ./dashboard_exports/:")
for f in sorted(os.listdir("dashboard_exports")):
    print(" -", f)
```

## Stage 8 — Roadmap: Aspect-Based Sentiment Analysis (not implemented here)

*"The food was amazing but traffic in KL was horrible"* currently gets **one** `final_sentiment` label for the
whole comment, and the aggregations above give it credit in both the `food` and `transport` aspect buckets under
that same single label — a comment-level approximation, not true phrase-level ABSA. This is workable for a first
version (it's why `tourism_aspect` is captured per comment already), but the next real upgrade is a proper ABSA
model that assigns a *separate* sentiment to each aspect mentioned in the same comment
(`food → positive, transport → negative`), which would directly sharpen the destination heatmap and the
top-negative-issues ranking above. Scoping and implementing that is a separate follow-up, not part of this
refinement.
